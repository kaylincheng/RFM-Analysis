```python
df=pd.read_csv("RFM_Analysis/RFM_Raw.csv")
df.head()
```

```python
import pandas as pd

# 1. 確保Date欄位為 datetime 格式（若你的 df 已經是則可省略此行）
df["Date"] = pd.to_datetime(df["Date"])

# 2. 找出整個資料集中的「最大Date」（作為計算 R 的基準點）
max_date = df["Date"].max()

# 3. 進行 Groupby 聚合計算
rfm_df = (
    df.groupby("CustomerKey")
    .agg(
        F=("InvoNo", "count"),  # InvoNo計數
        M=("Amount", "sum"),  # Amount加總
        # 計算該customer的最大Date，並與全體最大Date相減，最後轉換為天數
        R=("Date", lambda x: (max_date - x.max()).days),
    )
    .reset_index()
)

# 4. 重新命名欄位符合你的需求
rfm_df = rfm_df.rename(columns={"CustomerKey": "customer"})

# 調整欄位順序為：customer, F, M, R
rfm_df = rfm_df[["customer", "F", "M", "R"]]

# 查看結果
print(rfm_df)
```
